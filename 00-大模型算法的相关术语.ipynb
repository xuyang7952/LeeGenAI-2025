# 大模型的相关术语

## ​1. 核心模型相关​​
- ​​LLM​​ (Large Language Model): 大语言模型
- ​​LM​​ (Language Model): 语言模型
- ​​SLM​​ (Small Language Model): 小规模语言模型
- ​​FLM​​ (Foundation Language Model): 基础语言模型
- ​​MLLM​​ (Multimodal Large Language Model): 多模态大模型（如GPT-4V）


## 2. 技术方法​​
- ​​NLP​​ (Natural Language Processing): 自然语言处理
- ​​Transformer​​: 基于自注意力机制的模型架构（如GPT、BERT）
- ​​GPT​​ (Generative Pre-trained Transformer): OpenAI的生成式预训练模型
- ​​BERT​​ (Bidirectional Encoder Representations from Transformers): 双向编码器表示模型
- **​​RLHF​​ (Reinforcement Learning from Human Feedback): 人类反馈强化学习**
- **​​SFT​​ (Supervised Fine-Tuning): 监督微调**
- **​​PEFT​​ (Parameter-Efficient Fine-Tuning): 参数高效微调（如LoRA）**
- **​​LoRA​​ (Low-Rank Adaptation): 低秩适配微调技术**

## 3. 训练与优化​​
- **​​Pretraining​​: 预训练**
- **​post-training​​: 后训练**
- **​​Fine-tuning​​: 微调**
- ​​Zero-shot Learning​​: 零样本学习
- Few-shot Learning​​: 少样本学习
- **​​CoT​​ (Chain-of-Thought): 思维链提示**
- **​​MoE​​ (Mixture of Experts): 专家混合模型（如Google的Switch Transformer）**

## 4. 评估与安全​​
- ​​BLEU​​ (Bilingual Evaluation Understudy): 机器翻译评估指标
- ​​ROUGE​​ (Recall-Oriented Understudy for Gisting Evaluation): 文本摘要评估指标
- ​​Toxicity Detection​​: 毒性检测
- ​​AI Alignment​​: AI对齐（使模型目标与人类价值观一致）
- ​​Jailbreaking​​: 越狱（绕过模型安全限制的行为）

## 5. 硬件与计算​​
- ​​GPU​​ (Graphics Processing Unit): 图形处理器（常用于训练）
- ​​TPU​​ (Tensor Processing Unit): 谷歌张量处理器
- ​​HPC​​ (High-Performance Computing): 高性能计算
- ​​FLOPs​​ (Floating Point Operations): 浮点运算次数（衡量算力）

## ​6. 开源模型与工具​​
- ​​Hugging Face​​: 开源模型库和平台
- ​​LLaMA​​ (Large Language Model Meta AI): Meta开源的大模型
- ​​Alpaca/Vicuna​​: 基于LLaMA微调的模型
- ​​PyTorch/TensorFlow​​: 深度学习框架

## 7. 其他关键概念​​
- ​​AGI​​ (Artificial General Intelligence): 通用人工智能
- ​​API​​ (Application Programming Interface): 应用程序接口（如OpenAI API）
- ​**​RAG​​ (Retrieval-Augmented Generation): 检索增强生成**
- ​​KV Cache​​ (Key-Value Cache): 推理时的键值缓存优化技术
